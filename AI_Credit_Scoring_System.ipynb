```python
import matplotlib.pyplot as plt

# Data
years = ["2023", "2024"]
revenue = [6.8, 7]  # in millions USD
net_loss = [-502, -347]  # in millions USD

plt.figure(figsize=(8,5))
plt.bar(years, revenue, color="#4a90e2", label="Revenue (M USD)")
plt.bar(years, net_loss, color="#2c3e50", label="Net Loss (M USD)")
plt.title("Virgin Galactic: Annual Revenue and Net Loss")
plt.xlabel("Fiscal Year")
plt.ylabel("USD (Millions)")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```
